# Dimensionality Reduction using PCA and t-SNE

## Objective

To understand and demonstrate dimensionality reduction using:

1. Principal Component Analysis (PCA)
2. t-Distributed Stochastic Neighbor Embedding (t-SNE)

The project uses the MNIST handwritten digit dataset. Each image contains 28 × 28 pixels, resulting in 784 features. PCA and t-SNE are used to transform this high-dimensional data into a 2-dimensional representation for visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully!")
```

```python
import os

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
```

```python
# Dataset path
DATA_PATH = "/kaggle/input/datasets/hichamachahboun/mnist-handwritten-digits/"

# Load training data
X_train = np.load(DATA_PATH + "train_images.npy")
y_train = np.load(DATA_PATH + "train_labels.npy")

# Load testing data
X_test = np.load(DATA_PATH + "test_images.npy")
y_test = np.load(DATA_PATH + "test_labels.npy")

print("Training images shape:", X_train.shape)
print("Training labels shape:", y_train.shape)

print("Testing images shape:", X_test.shape)
print("Testing labels shape:", y_test.shape)
```

```python
# Display 10 sample images from the training dataset

plt.figure(figsize=(12, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(f"Label: {y_train[i]}")
    plt.axis("off")

plt.suptitle("Sample Images from MNIST Dataset", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Convert 28 × 28 images into 784-dimensional feature vectors

X_train_flat = X_train.reshape(X_train.shape[0], -1)
X_test_flat = X_test.reshape(X_test.shape[0], -1)

print("Original image shape:", X_train.shape)
print("Flattened training data shape:", X_train_flat.shape)
print("Flattened testing data shape:", X_test_flat.shape)
```

```python
# Normalize pixel values from 0-255 to 0-1

X_train_scaled = X_train_flat / 255.0
X_test_scaled = X_test_flat / 255.0

print("Minimum pixel value:", X_train_scaled.min())
print("Maximum pixel value:", X_train_scaled.max())
print("Scaled data shape:", X_train_scaled.shape)
```

```python
# Apply PCA to reduce 784 dimensions to 2 dimensions

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_train_scaled)

print("Original number of features:", X_train_scaled.shape[1])
print("Number of PCA components:", X_pca.shape[1])
print("PCA output shape:", X_pca.shape)
```

```python
# Calculate the variance captured by the two principal components

explained_variance = pca.explained_variance_ratio_

print("Variance explained by PC1:", explained_variance[0])
print("Variance explained by PC2:", explained_variance[1])
print("Total variance explained:", explained_variance.sum())
```

```python
# Visualize the PCA-reduced data

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=y_train,
    cmap="tab10",
    s=5,
    alpha=0.6
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization of MNIST Digits")

plt.colorbar(scatter, label="Digit")

plt.show()
```

```python
# Use a subset of 10,000 samples for faster t-SNE computation
n_samples = 10000

X_tsne_input = X_train_scaled[:n_samples]
y_tsne = y_train[:n_samples]

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42,
    init="pca",
    learning_rate="auto"
)

X_tsne = tsne.fit_transform(X_tsne_input)

print("t-SNE input shape:", X_tsne_input.shape)
print("t-SNE output shape:", X_tsne.shape)
```

```python
plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    c=y_tsne,
    cmap="tab10",
    s=5,
    alpha=0.6
)

plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.title("t-SNE Visualization of MNIST Digits")

plt.colorbar(scatter, label="Digit")

plt.show()
```

```python
# Calculate explained variance using all principal components
pca_full = PCA()
pca_full.fit(X_train_scaled)

cumulative_variance = np.cumsum(pca_full.explained_variance_ratio_)

plt.figure(figsize=(10, 6))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Cumulative Explained Variance")

plt.grid()
plt.show()
```

## Applications of PCA and t-SNE

### Applications of PCA

- **Data Compression:** Reduces the number of features while retaining important information.
- **Noise Reduction:** Removes less important variations and noise from high-dimensional data.
- **Data Visualization:** Converts high-dimensional data into 2D or 3D representations.
- **Feature Extraction:** Creates new features that capture the major patterns in the dataset.
- **Preprocessing:** Can reduce the number of features before applying machine learning algorithms.

### Applications of t-SNE

- **High-Dimensional Data Visualization:** Helps visualize complex datasets in 2D or 3D.
- **Cluster Exploration:** Makes groups of similar data points easier to observe.
- **Image Analysis:** Can visualize groups of similar images, such as handwritten digits.
- **Text and Embedding Visualization:** Helps visualize relationships between high-dimensional text or word embeddings.
- **Exploratory Data Analysis:** Helps researchers understand patterns and local relationships in complex datasets.

## Conclusion

In this project, dimensionality reduction was demonstrated using PCA and t-SNE on the MNIST handwritten digit dataset.

The original images contained 28 × 28 pixels, resulting in 784 features per image. PCA was used to transform the high-dimensional data into a 2-dimensional representation while preserving the directions with the highest variance. t-SNE was then used to create a 2-dimensional representation that focuses on preserving local similarities between data points.

The visualizations show that dimensionality reduction can make high-dimensional datasets easier to understand and analyze. PCA is useful for efficient dimensionality reduction, preprocessing, compression, and visualization, while t-SNE is particularly useful for exploring local patterns and clusters in high-dimensional data.

Overall, PCA and t-SNE provide useful approaches for understanding and visualizing complex high-dimensional datasets.
